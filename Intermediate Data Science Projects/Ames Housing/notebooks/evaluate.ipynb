## Machine Learning Conclusions

I trained four regression models to predict house sale price, namely, a dummy regressor, Support Vector Machine (SVR), Random Forest and XG Boost. Some conclusions worth noting are:
1.	The cross-validation and test-set R² scores are generally consistent, suggesting that the models generalize well and are not severely overfitting. The best test scores were achieved by Random Forest and XG Boost, both achieving 0.93. Ensemble models had the advantage over the other models, performing well overall and handling nonlinear data effectively.
2.	It is also worth noting that, in many cases, SVR performed worse than the dummy regressor. The SVR model failed to capture the relationship between the predictors and sale price. Its negative R² indicates that a naive strategy of always predicting the mean sale price outperformed the model.
The mean absolute error (MAE) is slightly better, getting its predictions wrong by about 58k, instead of 60k. 
3.	The top models were Random Forest and XG Boost, which excelled during testing. Both show a strong ability to learn the pattern effectively, achieving high test R2 scores and achieving mean absolute errors of ≈$13.8k. Furthermore, the low mean absolute percentage error (MAPE) of 8% suggests that their predictions are off by only 8%. Random Forest and XGBoost produced nearly identical results, with XGBoost achieving a marginally higher R² score. The difference is small enough that both models can be considered top performers on this dataset.
Tree-based ensemble methods substantially outperformed both the baseline and SVR models, suggesting that the relationship between housing features and sale price is highly nonlinear.

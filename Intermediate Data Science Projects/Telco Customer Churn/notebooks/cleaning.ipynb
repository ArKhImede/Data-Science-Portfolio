```python
import pandas as pd
from pathlib import Path

DF_PATH = Path("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")
df = pd.read_csv(DF_PATH)
```

```python
# look at the df
df.head()
```

```python
# shape of df
df.shape
```

```python
# info about the df
df.info(memory_usage="deep")
```

```python
# check duplicates
df.duplicated().sum()
```

```python
# check NaN values
df.isna().sum()
```

```python
# drop customer ID as we don't need it
df = df.drop("customerID", axis=1)
```

```python
# fix column name
df.rename(columns={"gender": "Gender", "tenure": "Tenure"}, inplace=True)
```

```python
# handle TotalCharges wrong data type
df.TotalCharges = pd.to_numeric(df.TotalCharges, errors="coerce")
```

```python
# check TotalCharges NaN values
df.TotalCharges.isna().sum()
```

```python
# fill them with 0
df.TotalCharges = df.TotalCharges.fillna(0)
```

```python
# change some str features to category for better performance
cols_to_change = ["Gender", "Partner", "Dependents", "PhoneService", "MultipleLines", "InternetService",
                  "OnlineSecurity", "OnlineBackup", "DeviceProtection", "TechSupport", "StreamingTV",
                  "StreamingMovies", "Contract", "PaperlessBilling", "PaymentMethod"]

for col_to_change in cols_to_change:
    df[col_to_change] = df[col_to_change].astype("category")
```

```python
# check unusual categories
for col_to_change in cols_to_change:
    print(df[col_to_change].value_counts())
```

```python
# check max. integer values to change data type (for better performance)
print(f"Min. SeniorCitizen: {df.SeniorCitizen.min()} | Max. SeniorCitizen: {df.SeniorCitizen.max()}")
print(f"Min. Tenure: {df.Tenure.min()} | Max. Tenure: {df.Tenure.max()}")
print(f"Min. MonthlyCharger: {df.MonthlyCharges.min()} | Max. MonthlyCharger: {df.MonthlyCharges.max()}")
```

```python
# change numerical data types
df.SeniorCitizen = df.SeniorCitizen.astype("int8")
df.Tenure = df.Tenure.astype("int8")
df.MonthlyCharges = df.MonthlyCharges.astype("float32")
```

```python
# look at the memory used
df.info(memory_usage="deep")
```

```python
# see final df
df.head()
```

```python
# save the cleaned dataframe
df.to_parquet("../data/processed/telco_customer_churn_cleaned.parquet", index=False)
```

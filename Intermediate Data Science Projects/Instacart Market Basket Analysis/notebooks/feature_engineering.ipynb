```python
import pandas as pd
from pathlib import Path
import sys
sys.path.append("../src")
from feature_engineering import convert_time_to_part_of_day, get_product_name_len

MASTER_DF_PATH = Path("../data/processed/master_df_cleaned.parquet")
master_df = pd.read_parquet(MASTER_DF_PATH)
```

```python
# see df
master_df.head()
```

## Creating New Features

```python
# morning/afternoon/evening buy
master_df["part_of_day_purchase"] = master_df.order_hour_of_day.apply(convert_time_to_part_of_day).astype("category")
```

```python
# product name length
master_df["product_name_length"] = master_df.product_name.apply(get_product_name_len).astype(int)
```

```python
# days since prior order missing
master_df["days_since_prior_order_missing"] = master_df.days_since_prior_order.isna().astype(int)
```

```python
# see final df
master_df.head()
```

```python
# save the feature engineered dataframe
master_df.to_parquet("../data/processed/master_df_feature_engineered.parquet", index=False)
```

```python
import pandas as pd
from pathlib import Path
import numpy as np

CHICAGO_CENSUS_PATH = Path("../data/raw/ChicagoCensusData.csv")
CHICAGO_CRIME_PATH = Path("../data/raw/ChicagoCrimeData.csv")
CHICAGO_PUBLIC_SCHOOLS_PATH = Path("../data/raw/ChicagoPublicSchools.csv")

chicago_census_df = pd.read_csv(CHICAGO_CENSUS_PATH)
chicago_crime_df = pd.read_csv(CHICAGO_CRIME_PATH)
chicago_public_schools_df = pd.read_csv(CHICAGO_PUBLIC_SCHOOLS_PATH)
```

# Chicago Public Schools Dataframe

```python
# look at the df
chicago_public_schools_df.head()
```

```python
# shape
chicago_public_schools_df.shape
```

```python
# info
chicago_public_schools_df.info(memory_usage="deep")
```

```python
# check duplicates
chicago_public_schools_df.duplicated().sum()
```

```python
# check NaN values
chicago_public_schools_df.isna().sum().sort_values(ascending=False)
```

```python
# replace NDA with NaN
chicago_public_schools_df.replace("NDA", np.nan, inplace=True)
```

```python
# drop Location since we already have Latitude and Longitude
chicago_public_schools_df = chicago_public_schools_df.drop("Location", axis=1)
```

```python
# replace % with "" in columns that contain %
perc_cols = ["AVERAGE_STUDENT_ATTENDANCE", "Average_Teacher_Attendance", "Individualized_Education_Program_Compliance_Rate"]

for perc_col in perc_cols:
    chicago_public_schools_df[perc_col] = chicago_public_schools_df[perc_col].astype("str").str.replace("%", "", regex=False)
    chicago_public_schools_df[perc_col] = chicago_public_schools_df[perc_col].astype(float)
```

```python
# drop columns ending with _Icon as other columns contain more information
icon_cols = [c for c in chicago_public_schools_df.columns if "_Icon" in c]
chicago_public_schools_df.drop(columns=icon_cols, inplace=True)
```

```python
# convert needed columns for clustering to numeric
num_cols = ["SAFETY_SCORE", "Family_Involvement_Score", "Environment_Score", 
            "Instruction_Score", "AVERAGE_STUDENT_ATTENDANCE"]

for num_col in num_cols:
    chicago_public_schools_df[num_col] = pd.to_numeric(chicago_public_schools_df[num_col], errors="coerce")
```

```python
# NaN summary
chicago_public_schools_df.isna().mean().sort_values(ascending=False)
```

```python
# info about the df
chicago_public_schools_df.info(memory_usage="deep")
```

```python
# aggregate public school data
public_school_agg = chicago_public_schools_df.groupby("COMMUNITY_AREA_NUMBER").agg(
    UniqueNetworkManagersCount = ("Network_Manager", "nunique"),
    MeanSafetyScore = ("SAFETY_SCORE", "mean"),
    MeanFamilyInvolvementScore = ("Family_Involvement_Score", "mean"),
    MeanEnvironmentScore = ("Environment_Score", "mean"),
    MeanInstructionScore = ("Instruction_Score", "mean"),
    MeanStudentAttendance = ("AVERAGE_STUDENT_ATTENDANCE", "mean"),
).reset_index()

# show it
public_school_agg.head()
```

# Chicago Crime Dataframe

```python
# look at the df
chicago_crime_df.head()
```

```python
# shape
chicago_crime_df.shape
```

```python
# info
chicago_crime_df.info(memory_usage="deep")
```

```python
# check duplicates
chicago_crime_df.duplicated().sum()
```

```python
# check NaN values
chicago_crime_df.isna().sum()
```

```python
# since some community area numbers (43) are missing, we can drop them
chicago_crime_df.dropna(subset=["COMMUNITY_AREA_NUMBER"], inplace=True)
```

```python
# aggregate
crime_agg = chicago_crime_df.groupby("COMMUNITY_AREA_NUMBER").agg(
    TotalCrimes = ("ID", "count"),
    UniqueCrimeCategories = ("PRIMARY_TYPE", "nunique"),
    ArrestsCount = ("ARREST", "sum"),
).reset_index()

# show it
crime_agg.head()
```

```python
# merge all aggregated dataframes
merged = chicago_census_df.merge(
    public_school_agg, how="left", on="COMMUNITY_AREA_NUMBER"
    ).merge(crime_agg, how="left", on="COMMUNITY_AREA_NUMBER")
```

```python
# look at the merged df
merged.head()
```

```python
# info about the merged df
merged.info(memory_usage="deep")
```

```python
# since no crime records usually means no crimes, we can fill those features with 0
crime_cols = ["TotalCrimes", "UniqueCrimeCategories", "ArrestsCount"]
merged[crime_cols] = merged[crime_cols].fillna(0)
```

```python
# check NaN values
merged.isna().sum()
```

```python
# use median for quality features
quality_cols = ["MeanSafetyScore", "MeanFamilyInvolvementScore", "MeanEnvironmentScore", "MeanInstructionScore"]

for col in quality_cols:
    merged[col] = merged[col].fillna(merged[col].median())
```

```python
# check that NaN values are 0 after merging and imputing
assert merged.isna().sum().sum() == 0
```

```python
# check if there are duplicates
assert merged.duplicated().sum() == 0
```

```python
# info about merged
merged.info(memory_usage="deep")
```

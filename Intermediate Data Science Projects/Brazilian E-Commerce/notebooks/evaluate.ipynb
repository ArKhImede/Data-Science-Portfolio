## Machine Learning Conclusions

1. Overall, all of the three regression models performed poorly across every evaluation score.
2. The available features contain limited predictive information about review scores as evidenced by the poor r² score, which is low, close to 0. SVR generated a negative score, meaning it performed worse than simply predicting the average.
3. The MAE scores seem low, but actually they aren't. On average, predictions are off by 1, which is high for review scores (range 1-5).
4. Point 3 is further emphasized by the third metric, MSE, which penalizes large model mistakes. For review scores, the metric isn't meaningful as squared review scores don't make much sense, but still confirms weak performance because large prediction errors are penalized more heavily.

Since review scores are ordinal values from 1 to 5, future work could explore classification approaches instead of regression.

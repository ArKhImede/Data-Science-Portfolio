```python
import pandas as pd
from pathlib import Path

MAIN_PATH = Path("../data/processed/main_df.csv")
ORDER_ITEMS_AND_PRODUCTS_PATH = Path("../data/processed/order_items_and_products_df.csv")

order_items_and_products_df = pd.read_csv(ORDER_ITEMS_AND_PRODUCTS_PATH)
```

```python
# order items and products df
order_items_and_products_df.head()
```

```python
# total spend
order_items_and_products_df["total_spend"] = order_items_and_products_df.price + order_items_and_products_df.freight_value
```

```python
# price bins
order_items_and_products_df.price.value_counts(bins=5)
price_labels = ["Low", "Medium", "High"]
price_bins = [0, 50, 500, 10000]

order_items_and_products_df["price_category"] = pd.cut(order_items_and_products_df.price, bins=price_bins, labels=price_labels)
```

```python
# volume 
order_items_and_products_df["volume"] = order_items_and_products_df.product_length_cm * order_items_and_products_df.product_height_cm * order_items_and_products_df.product_width_cm
```

```python
# density
order_items_and_products_df["density"] = (order_items_and_products_df.product_weight_g / order_items_and_products_df.volume).astype(float).round(2)
```

```python
# freight per price ratio
order_items_and_products_df["freight_per_price"] = (order_items_and_products_df.freight_value / order_items_and_products_df.price).astype(float).round(2)
```

```python
# final dataframe
order_items_and_products_df.head()
```

```python
# save feature engineered dataset
order_items_and_products_df.to_csv("../data/processed/order_items_and_products_feature_engineered_df.csv", index=False)
```

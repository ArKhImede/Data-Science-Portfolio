```python
import pandas as pd
from pathlib import Path
from src.preprocessing import (convert_date_columns, 
                           convert_response_rate, 
                           convert_boolean_columns, 
                           drop_unused_columns, 
                           handle_numeric_missing_values, 
                           handle_categorical_missing_values)

DF_PATH = Path("../data/raw/Airbnb_Data.csv")
df = pd.read_csv(DF_PATH)
```

```python
# look at the DataFrame
df.head()
```

```python
# shape
df.shape
```

## Data Type Conversion

```python
# convert date columns
df = convert_date_columns(df)

# convert response rate
df = convert_response_rate(df)

# convert boolean columns
df = convert_boolean_columns(df)
```

## Remove Unnecessary Columns

```python
# drop unused columns
df = drop_unused_columns(df)
```

## Handle Missing Values

```python
# handle numeric missing values
df = handle_numeric_missing_values(df)

# handle categorical missing values
df = handle_categorical_missing_values(df)
```

Date-related missing values are intentionally preserved. They are informative and will be handled during feature engineering.

## Check final DataFrame

```python
# Verify remaining missing values
df.isna().sum()
```

```python
# move target (log_price) to the end
col_to_move = df.pop("log_price")
df.insert(len(df.columns), "log_price", col_to_move)
```
